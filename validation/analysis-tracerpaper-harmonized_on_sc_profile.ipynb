# Paper figures on the harmonized runs ("Harmonization B", every storm on the supercell's profile)

This is the manuscript's analysis notebook — "The Entrainment of Air from Rainy Surface Regions and its Implications for Bioaerosol Transport in Deep Convection" — pointed at the harmonized set of runs instead of the runs the manuscript was written from.

## What this notebook is

Every analysis and figure cell below is a byte-for-byte copy of
`replication/analysis-tracerpaper.ipynb`. Only the cells that decide *which simulation
runs are read*, *where results are written*, and the one calculation that the harmonized
runs cannot support have been changed. That is deliberate. The comparison between this
notebook's figures and the manuscript's is only meaningful if the analysis code is
identical and the input runs are the only thing that differs.

## The runs it reads

"Harmonization B" puts all three storms on the supercell's (correct) initial
thermodynamic profile:

| Storm | Run | Same as the manuscript's? |
| --- | --- | --- |
| Isolated convection | `from_cumulus/ic_wk/validation/sc_profile` | No — rerun |
| Squall line | `from_cumulus/sl_wk/sc_profile` | No — rerun |
| Supercell | `supercell_wk/all_tracers/newtracer_forcing-grant2014_dx-500_frqlite-2_stratwinds` | Yes — unchanged |

The registry and the provenance audit for those runs live in
`validation/harmonization_utils.py`, under `HARMONIZED_ON_SC_PROFILE_RUN_DIRS`. The
side-by-side comparison of this arm against the manuscript's arm lives in
`validation/harmonized_on_sc_profile.ipynb`. This notebook is the complement to that
one: it produces the manuscript's own figures, unmodified, for this arm alone.

## Every way this notebook differs from the manuscript's

1. **Imports.** `harmonization_utils` is imported first, because importing it is what
   puts the in-repo `cloudy` and `blt_utils_limited` packages on the path. This notebook
   lives in `validation/` rather than `replication/`, so neither is importable by
   default.

2. **Run selection.** A new cell, "Arm configuration", overwrites `blt_utils`'s
   module-level path constants in place. The manuscript notebook selects its runs by
   *being* that module; this one cannot, so it repoints the constants after import.
   Several cells call `reload(blt)`, which restores the manuscript's values, so each of
   those cells now calls `point_blt_at_this_arm()` immediately afterwards.

3. **Output location.** Figures and figure data go to
   `validation/harmonization_output/harmonized-sc-profile/paper_figures/` and
   `.../paper_figure_data/`, never to `paper_figures/`. A run of this notebook must not
   be able to overwrite a manuscript figure.

4. **Derived-product caching.** The tracer profile datasets and development dataframes
   are loaded through `harmonization_utils`, which reads a cache only when its number of
   output times matches the number of `a-A-*.h5` files the run actually has, and
   recomputes it otherwise. The manuscript notebook uses a hand-set per-storm flag
   instead. That is not good enough here, because a rerun's cache can be written while
   the run is still integrating, and a truncated cache read back is invisible
   downstream — every figure would quietly describe a shorter simulation.

5. **Upper-level mass flux (Figures 11 and 12).** This is the one calculation that
   *cannot* be reproduced as the manuscript does it. The manuscript integrates the flux
   from the 2-second "lite" output (`a-L-*.h5`). Both reruns in this arm were submitted
   with `FRQLITE = 0` and wrote no lite output at all, so those files do not exist.
   A 5-minute estimator is substituted, applied identically to all three storms. It is
   biased, and section "Figures 11-12" quantifies the bias on the supercell — the one
   storm here that does have lite output. **Absolute flux numbers in Figures 11 and 12
   are therefore not comparable to the manuscript's; ratios between storms are, to the
   extent the bias is common-mode.**

6. **Sensitivity tests.** That section was already stale in the source notebook: four
   of its cells reference variables (`ul_flux_dss`, `cumu_upward_ul_mass_fluxes`,
   `lite_dss`) that no longer exist anywhere in it, and would raise `NameError` on a
   top-to-bottom run. The names are repaired here so the notebook completes. See the
   note at the head of that section for what changed and what it now means.

## Running it

Run top to bottom. `USE_DASK_LOCAL_CLUSTER` is `True`; the reductions want a cluster.

Every derived product this notebook needs is either already on disk or computed and
cached on first use, so no preprocessing step is required. The two reruns already have
their tracer profile datasets, development dataframes and upper-level tracer totals
cached, so the expensive first pass has already been paid for them.

## Arm configuration

Everything that makes this notebook read the harmonized runs rather than the manuscript's runs is in the cell below. Nothing else in the notebook names a run directory.

```python
# The arm this notebook analyses. "Harmonization B" puts every storm on the
# supercell's (correct) initial profile: the isolated convection and the squall line were
# rerun on that profile, and the supercell is unchanged, so for the supercell this is the
# same run the manuscript used.
ARM_LABEL = "harmonized-sc-profile"
RUN_DIRS = hu.HARMONIZED_ON_SC_PROFILE_RUN_DIRS

# Where this notebook writes figures and figure data.
#
# Deliberately NOT blt.FIGS_DIR. That is `paper_figures/`, and a run of this notebook
# must never be able to overwrite a manuscript figure with a harmonized-arm one carrying
# the same filename. The directory names are also distinct from the ones
# `harmonized_on_sc_profile.ipynb` writes to (`figures/` and `figure_data/`), so the two
# notebooks cannot overwrite each other either.
ARM_OUTPUT_DIR = hu.HARMONIZATION_OUTPUT_ROOT / ARM_LABEL
FIGURE_OUTPUT_DIR = ARM_OUTPUT_DIR / "paper_figures"
FIGURE_DATA_OUTPUT_DIR = ARM_OUTPUT_DIR / "paper_figure_data"

# Arm-level cache directory for the derived products that `blt` keys by threshold height
# alone and not by run. `blt.DERIVED_DIR` normally points into the in-repo paper tree, so
# leaving it alone would make this notebook read the manuscript's caches and silently
# report the manuscript's numbers under this notebook's name.
DERIVED_OUTPUT_DIR = ARM_OUTPUT_DIR / "paper_derived"

for _output_dir in [
    FIGURE_OUTPUT_DIR,
    FIGURE_DATA_OUTPUT_DIR,
    DERIVED_OUTPUT_DIR,
]:
    _output_dir.mkdir(parents=True, exist_ok=True)


def point_blt_at_this_arm():
    """Repoint every run-specific path constant in ``blt`` at this notebook's arm.

    Mechanism: ``blt_utils.py`` hard-codes one run directory per storm as module-level
    constants, so the manuscript notebook selects its runs by *being* that module. This
    notebook analyses different runs, so it overwrites those constants in place after
    importing the module.

    The per-storm paths are built with ``hu``'s path helpers rather than written out as
    literals, because the paper runs and the reruns do not keep their derived products in
    the same place. Each rerun keeps its own beside the run, in ``<run_dir>/derived/``;
    the paper runs have no populated ``derived/`` directory at all and keep theirs in the
    in-repo tree ``replication/data/derived/<storm>/``. This arm needs both conventions at
    once, because its supercell is a paper run and its other two storms are reruns.
    Getting that wrong would not raise. The affected storm would simply find no cache and
    recompute every derived product from raw output, taking hours and writing the results
    into the wrong tree.

    Why this is a function rather than a block of assignments: several cells below call
    ``reload(blt)``, which re-executes the module body and restores every one of these
    constants to the manuscript's values. Any cell that reloads therefore has to call this
    immediately afterwards. That failure would also be silent -- the notebook would read
    the manuscript's runs and reproduce the manuscript's figures under this notebook's
    name.
    """
    # The working copy of blt_utils.py has STORM_NAMES trimmed to one storm for an
    # unrelated debugging session; this notebook always wants all three.
    blt.STORM_NAMES = list(hu.STORM_KEYS_IN_FIGURE_ORDER)

    # Raw 5-minute analysis output.
    blt.RAMS_OUTPUT_DIRS = {
        storm_key: hu.analysis_output_dir(run_dir)
        for storm_key, run_dir in RUN_DIRS.items()
    }
    # In-updraft tracer vertical profiles (Figures 7 and 9, and the entrainment stats).
    blt.TRACER_TOTALS_PROFILE_FILEPATHS = {
        storm_key: hu.tracer_totals_profile_filepath(run_dir, storm_key)
        for storm_key, run_dir in RUN_DIRS.items()
    }
    # Storm development dataframes (Figure 10).
    blt.DEVELOPMENT_DATAFRAME_FILEPATHS = {
        storm_key: hu.development_dataframe_filepath(run_dir, storm_key)
        for storm_key, run_dir in RUN_DIRS.items()
    }

    # Figure and figure-data output.
    blt.FIGS_DIR = FIGURE_OUTPUT_DIR
    blt.FIGURE_DATA_DIR = FIGURE_DATA_OUTPUT_DIR

    # Upper-level caches. UL_FLUXES_DIR and UL_TRACER_TOTALS_FILEPATHS are derived from
    # DERIVED_DIR when the module is imported, so they have to be rebuilt here rather than
    # left to follow the new value of DERIVED_DIR on their own.
    blt.DERIVED_DIR = DERIVED_OUTPUT_DIR
    blt.UL_FLUXES_DIR = DERIVED_OUTPUT_DIR
    blt.UL_TRACER_TOTALS_FILEPATHS = {
        storm_key: (
            DERIVED_OUTPUT_DIR
            / storm_key
            / f"ul_tracer_totals_ds_z-{blt.UL_Z_THRESHOLD}.pkl"
        )
        for storm_key in blt.STORM_NAMES
    }


point_blt_at_this_arm()
```

```python
reload(cm.rams)
reload(blt)
# reload(blt) restores blt_utils.py's own path constants, which point at the
# manuscript's runs. Repoint them before anything below reads a file.
point_blt_at_this_arm()

# Define the list of RAMS output filenames we'll read in
# Can specify this manually or just use a glob
rams_input_filename_glob = "a-A-*.h5"

rams_input_dim_names_kwargs = (
    {
        "dim_names": {
            "phony_dim_0": "y",
            "phony_dim_1": "x",
            "phony_dim_2": "p",
            "phony_dim_3": "z",
        }
    }
    if rams_input_filename_glob.startswith("a-L")
    else {}
)

storm_dss = {}

for storm_name in tqdm(list(blt.STORM_LABELS.keys())):
    print(blt.RAMS_OUTPUT_DIRS[storm_name])
    rams_input_filenames = sorted(
        list(blt.RAMS_OUTPUT_DIRS[storm_name].glob(rams_input_filename_glob))
    )

    # Read in the RAMS output
    rams_ds = cm.rams.read_rams_output(
        input_filenames=rams_input_filenames,
        time_dim_name="time",
        units=False,
    )

    # Postprocess
    rams_ds = blt.postprocess_storm_ds(rams_ds, storm_name)

    # Store
    storm_dss[storm_name] = rams_ds


# Calculate the timestep and horizontal grid spacing, assuming they are uniform
# Doesn't matter which storm ds we use
DELTA_T = (
    rams_ds["time"].values[1] - rams_ds["time"].values[0]
) / np.timedelta64(1, "s")
DELTA_X = rams_ds["x"].values[1] - rams_ds["x"].values[0]
```

## Figure 1: Storm soundings

```python
reload(blt)
# reload(blt) restores the manuscript's path constants; put this arm's back.
point_blt_at_this_arm()

figure_name = "f01_skewt_logp"
figure_data = {}

storm_skewt_dfs = {
    storm: (
        storm_ds.isel({"time": 0})
        .mean(["x", "y"])[["P", "T", "dewpoint", "UC", "VC"]]
        .to_pandas()
        .reset_index()
    )
    for storm, storm_ds in storm_dss.items()
}

import matplotlib.ticker as ticker

fig = plt.figure(figsize=(10, 5), layout="constrained")
subfigs = fig.subfigures(ncols=3)


def convert_np(arr, ufrom, uto):
    return (arr * units(ufrom)).to(uto).magnitude


legend_artists = []

for storm_ix, (storm, this_ul_lite_ds) in enumerate(storm_skewt_dfs.items()):

    subfig = subfigs[storm_ix]
    wk_skewt = SkewT(subfig, rotation=30, aspect=1 * 100)
    t_line = blt.plot_and_save(
        figure_data,
        f"storm-{storm}_skewt_temperature",
        wk_skewt.plot,
        this_ul_lite_ds["P"].values,
        convert_np(this_ul_lite_ds["T"].values, "K", "degC"),
        "r",
        label="Temperature",
    )
    dp_line = blt.plot_and_save(
        figure_data,
        f"storm-{storm}_skewt_dewpoint",
        wk_skewt.plot,
        this_ul_lite_ds["P"].values,
        convert_np(this_ul_lite_ds["dewpoint"].values, "K", "degC"),
        "blue",
        label="Dewpoint",
    )

    # Calculate and plot parcel profile
    wk_parcel_path = mpc.parcel_profile(
        this_ul_lite_ds["P"].values * units("hPa"),
        this_ul_lite_ds["T"].values[0] * units("K"),
        this_ul_lite_ds["dewpoint"].values[0] * units("K"),
    )
    parcel_line = blt.plot_and_save(
        figure_data,
        f"storm-{storm}_surface_parcel_path",
        wk_skewt.plot,
        this_ul_lite_ds["P"].values,
        wk_parcel_path,
        color="grey",
        linestyle="dashed",
        linewidth=2,
        label="Surface parcel path",
    )
    # subfig.suptitle(f"$U_S$={Us} m/s, $Q_{{v0}}$={qv0} g/kg, shear scale height={zscale} m")

    # Add a hodograph
    from mpl_toolkits.axes_grid1.inset_locator import inset_axes

    skewt_ax = wk_skewt.ax
    skewt_ax.set_xlabel("Temperature (°C)")
    skewt_ax.set_ylabel("Pressure (hPa)")
    skewt_ax.yaxis.set_major_locator(ticker.MultipleLocator(200))
    for label in skewt_ax.xaxis.get_ticklabels()[1::2]:
        label.set_visible(False)

    ax_hod = inset_axes(skewt_ax, "30%", "30%", loc=1)
    h = Hodograph(ax_hod, component_range=35.0)
    h.add_grid(increment=10)
    # Only add the winds up to 4 km, so we can see the variation in hodograph color
    this_hodo_ds = this_ul_lite_ds.loc[this_ul_lite_ds["z"] <= 8000]
    blt.plot_and_save(
        figure_data,
        f"storm-{storm}_hogograph",
        h.plot_colormapped,
        this_hodo_ds["UC"].values,
        this_hodo_ds["VC"].values,
        this_hodo_ds["z"].values,
    )
    ax_hod.set_ylabel("U (m/s)")
    ax_hod.set_xlabel("V (m/s)")

    skewt_ax.set_title(blt.axes_title(storm_ix, blt.STORM_LABELS[storm]))
    # Save the legend artists; this just gets overwritten every time, which is fine
    legend_artists[:] = [t_line[0], dp_line[0], parcel_line[0]]

fig.legend(handles=legend_artists, loc="lower center", ncols=3)

blt.save_figure_data(figure_data, figure_name)
cm.plotting.fig_multisave(fig, figure_name, blt.FIGS_DIR)
```

```python
cm.plotting.start_figure("f06_surface_inp_concentrations")


# Read in slope of fits from Claudia's paper
mignani_slopes = pd.read_csv(
    "/moonbow/cmdavis4/projects/bl_transport/251006_INPconc-SumP_slope_intercept_results_forCharles.csv"
)


# Make a function to convert from a cumulative precipitation to an INP concentration
def p_to_inp(cumu_p, inp_temp):
    # Set the minimum and maximum values beyond which we don't extrapolate
    min_p = 0.001  # mm
    max_p = 9.557  # mm
    # Get the row for this inp temp
    this_line_params = mignani_slopes[
        mignani_slopes["Temperature.C"] == inp_temp
    ].iloc[0]
    # Fix where we would be extrapolating
    cumu_p = cumu_p.where(cumu_p < max_p, max_p)
    # print(this_line_params["slope"])

    # Now do the power law
    # This result in is in #/L
    return 10 ** (this_line_params["intercept"]) * cumu_p.where(
        cumu_p >= min_p, 0
    ) ** (this_line_params["slope"])


display(mignani_slopes)

this_fig_freezing_temps = [-8, -10]

fig, axs = plt.subplots(
    ncols=3,
    nrows=len(this_fig_freezing_temps),
    figsize=(8, 2.2 * len(this_fig_freezing_temps)),
    sharex="col",
    sharey="col",
)

ax_lims = {
    "ic_wk": {"x": [85000, 155000], "y": [85000, 135000]},
    "sl_wk": {"x": [60000, 225000], "y": None},
    "sc_wk": {"x": [75000, 300000], "y": [50000, 200000]},
}


def lm_polygon():
    return plt.Polygon(
        [
            (
                storm_ds["x"].min(),
                blt.sc_wk_right_mover_boundary(storm_ds["x"].min()),
            ),
            (storm_ds["x"].min(), storm_ds["y"].max()),
            (storm_ds["x"].max(), storm_ds["y"].max()),
            (
                storm_ds["x"].max(),
                blt.sc_wk_right_mover_boundary(storm_ds["x"].max()),
            ),
        ],
        closed=True,
        facecolor=(1, 1, 1, 0.8),
        edgecolor="tomato",
        linestyle="dashed",
        linewidth=2,
        zorder=999,
    )


for ax in axs[-1, :]:
    ax.set_xlabel("x (km)")

during_rain_times = {
    k: 5 * development_df["total_rain"].argmax()
    for k, development_df in development_dfs.items()
}
print(during_rain_times)

# Precalculate the values
n_inps = {}
for freezing_temp in this_fig_freezing_temps:
    this_temp_n_inps = {}
    for storm_ix, (storm, storm_ds) in enumerate(storm_dss.items()):
        this_temp_n_inps[storm] = p_to_inp(
            storm_ds["ACCPR"].sel({
                "time": blt.minutes_to_simulation_time(during_rain_times[storm])
            }),
            freezing_temp,
        )
    n_inps[freezing_temp] = this_temp_n_inps
# display(n_inps[-10]['ic_wk'])


for storm_ix, (storm, storm_ds) in enumerate(storm_dss.items()):
    axs[0, storm_ix].set_title(blt.STORM_LABELS[storm])
    for freezing_temp_ix, freezing_temp in enumerate(this_fig_freezing_temps):
        this_inp_ds = n_inps[freezing_temp][storm]
        ax = axs[freezing_temp_ix, storm_ix]
        pc = ax.pcolormesh(
            storm_ds["x"],
            storm_ds["y"],
            this_inp_ds,
            norm=colors.LogNorm(vmin=1e-2, vmax=4e-1),
            name=f"storm-{storm}_freezingtemp-{freezing_temp}_inps",
        )

        # Calculate the median and add it as text
        this_median = (
            this_inp_ds.where((this_inp_ds > 0) & (storm_ds["in_analysis"]))
            .compute()
            .median()
        )
        ax.text(
            0.02,
            0.02,
            rf"Median = {this_median:.2f} $L^{{-1}}$ STP",
            transform=ax.transAxes,
            va="bottom",
            ha="left",
            bbox=dict(boxstyle="round", facecolor="white", alpha=0.8),
        )

        # cbar = plt.colorbar(pc, ax=ax)
        # cbar.ax.set_ylabel("#/L")
        ax.set_xlim(ax_lims[storm]["x"])
        ax.set_ylim(ax_lims[storm]["y"])
        if storm == "sc_wk":
            ax.add_patch(lm_polygon())
        cm.plotting.scale_axes_ticks(ax)
        if storm_ix == 0:
            ax.set_ylabel("y (km)")
            cm.plotting.add_row_header(
                ax, rf"$T_{{freeze}}$ = {freezing_temp} °C"
            )
cb = fig.colorbar(pc, ax=axs, fraction=0.1)
cb.ax.set_ylabel(r"Estimated INP concentration ($L^{-1}$ STP)")
save_figure_and_data()
```

## Figures 11-12: Lofting to upper levels

### Setup

#### Calculate upper-level tracer concentrations

```python
# Total tracer content above the upper-level threshold height, as a time series.
#
# Two changes from the manuscript notebook, both about caching rather than about the
# calculation:
#
# 1. The reduction itself now comes from `hu.calculate_upper_level_tracer_dataframe`
#    rather than from a copy of it defined here. The two are the same calculation. Keeping
#    a local copy would mean a cache file on disk whose producer lives somewhere else,
#    and no way to notice if the two drifted apart.
#
# 2. The cache is per run, under that run's own `derived/upper_level/`, via
#    `hu.read_or_cache_dataframe`. The manuscript notebook caches one dict-of-all-storms
#    per threshold height under `blt.DERIVED_DIR`, a path that carries no run identity.
#    It also caches it with `cm.utils.read_or_cache_to`, which cannot in fact write `.pkl`
#    files at all: `cm.utils.write_file` dispatches `.pkl` to a `to_pkl` attribute that no
#    dict or DataFrame has, raises, and has that swallowed by `fail_ok=True`. The result
#    is a cache that silently never populates while printing that it wrote the file.
#
# `expected_row_count` is what makes a cache written mid-run self-correcting: one row per
# output time is the shape of this dataframe, so a different length means the cache
# predates the run finishing and must be redone.
#
# As in the manuscript, this is deliberately NOT restricted to the supercell's right
# mover -- the upper-level totals are for the full domain.
UPPER_LEVEL_Z_THRESHOLD_M = blt.UL_Z_THRESHOLD * 1000

ul_tracer_dfs_all_heights = {}
for z_km in blt.ALL_UL_Z_THRESHOlDS:
    this_z_ul_tracer_dfs = {}
    for storm, this_storm_ds in tqdm(storm_dss.items()):
        cache_dir = hu.upper_level_derived_dir(RUN_DIRS[storm], storm)
        cache_dir.mkdir(parents=True, exist_ok=True)
        this_z_ul_tracer_dfs[storm] = hu.read_or_cache_dataframe(
            cache_dir / f"ul_tracer_totals_z-{z_km * 1000}.pkl",
            hu.calculate_upper_level_tracer_dataframe,
            this_storm_ds,
            z_km * 1000,
            expected_row_count=this_storm_ds.sizes["time"],
        )
    ul_tracer_dfs_all_heights[z_km] = this_z_ul_tracer_dfs
```

#### Calculate upper level mass fluxes

```python
# Cumulative upward and downward mass flux through the upper-level threshold height.
#
# This is the one calculation in this notebook that cannot be done the way the manuscript
# does it. The manuscript integrates the flux from the 2-second "lite" output files
# (`a-L-*.h5`), sampling the vertical velocity field 150 times per analysis interval. Both
# reruns in this arm were submitted with `FRQLITE = 0` and wrote no lite output at all, so
# those files do not exist and cannot be recovered short of rerunning the simulations.
#
# The substitute samples the same quantity at the 5-minute analysis cadence instead: each
# output time contributes `w * rho * dx^2 * dt` with `dt` the analysis interval, split
# into upward and downward parts by the sign of w, and accumulated in time. Air density
# comes from P, T and mixing ratio rather than from DN0, matching how the manuscript's
# lite-file path does it (lite files carry no DN0), so the two estimators differ only in
# how often they sample.
#
# It is a biased estimator and knowingly so: 5-minute sampling aliases the
# convective-scale vertical velocity field badly, and there is no reason to expect it to
# recover the 2-second answer. It is applied identically to all three storms here, which
# is what keeps the storm-to-storm comparison meaningful. The next cell measures the bias
# directly on the supercell -- the one storm in this arm whose run does have lite output.
#
# What this means for reading the figures: every absolute flux number in Figures 11 and 12
# below differs from the manuscript's, and the quantities Figure 11 normalizes by flux
# shift with it. Ratios between storms survive to the extent that the bias is common-mode.
UPPER_LEVEL_Z_THRESHOLD_M = blt.UL_Z_THRESHOLD * 1000


def calculate_ul_flux_dss(ul_z_threshold):
    """Coarse (5-minute) cumulative up/down mass flux through one level, per storm."""
    return {
        storm: (
            hu.calculate_cumulative_mass_flux_from_analysis_output(
                this_storm_ds,
                upper_level_z_threshold_m=ul_z_threshold,
                output_interval_seconds=DELTA_T,
                horizontal_grid_spacing_m=DELTA_X,
            ).compute()
        )
        for storm, this_storm_ds in tqdm(storm_dss.items())
    }


# No caching here, unlike the manuscript's version. Reading 5401 lite files per storm was
# what made that worth caching; this reduction touches one model level at 37 output times
# and takes seconds.
ul_flux_dss_all_heights = {
    z_km: calculate_ul_flux_dss(z_km * 1000) for z_km in blt.ALL_UL_Z_THRESHOlDS
}
```

```python
# How wrong is the 5-minute flux estimator?
#
# The supercell is the same run in this arm as in the manuscript, and it has the 2-second
# lite output, so its exact flux is already cached in the in-repo paper tree. Comparing
# the two estimators on that one storm puts a number on the bias the two reruns' figures
# inherit.
#
# Read `blt.DERIVED_DIR` is NOT used here: the arm configuration cell repointed it at this
# arm's cache directory, and the file wanted is the manuscript's. Name the paper tree
# explicitly instead.
exact_flux_cache_path = (
    hu.PAPER_DERIVED_ROOT / f"ul_mass_fluxes_z-{blt.UL_Z_THRESHOLD}000.pkl"
)
print(
    f"Reading the manuscript's exact (2 s) fluxes from {exact_flux_cache_path}"
)
exact_flux_dss = pd.read_pickle(exact_flux_cache_path)

flux_validation_rows = []
for storm in blt.STORM_NAMES:
    # Only storms whose run is unchanged from the manuscript can be compared this way.
    # For the two reruns the cached exact flux belongs to a different simulation, so the
    # comparison would measure the harmonization rather than the estimator.
    if RUN_DIRS[storm] != hu.PAPER_RUN_DIRS[storm]:
        continue
    coarse_upward = float(
        ul_flux_dss_all_heights[blt.UL_Z_THRESHOLD][storm][
            "cumu_upward_mass_flux"
        ]
        .isel(time=-1)
        .values
    )
    exact_upward = float(
        exact_flux_dss[storm]["cumu_upward_mass_flux"].isel(time=-1).values
    )
    flux_validation_rows.append({
        "storm": blt.STORM_LABELS[storm],
        "exact_2s_upward_kg": exact_upward,
        "coarse_5min_upward_kg": coarse_upward,
        "coarse_over_exact": coarse_upward / exact_upward,
    })

display(
    HTML(
        "<h4>Coarse (5 min) vs exact (2 s) cumulative upward mass flux at the"
        " end of the simulation</h4>"
    )
)
display(pd.DataFrame(flux_validation_rows).set_index("storm"))
print(
    "Only storms whose run is unchanged from the manuscript appear above. For a"
    " storm-by-storm version of this check across all three, see the 'Figures"
    " 11-12' section of harmonized_on_sc_profile.ipynb, which computes both"
    " estimators on all three paper runs."
)
```

## Tracer entrainment statistics

```python
# Set the number of decimal places in the pandas display
# pd.options.display.float_format = "{:,.3f}".format
# pd.set_option('display.precision', 3)

tracer_stats = {}

tracer_stats_tracers = blt.RS_TRACERS + ["total_fs_tracer"]

for storm_ix, (storm, this_tracer_ds) in enumerate(tracer_totals_dss.items()):
    # Limit to only when there's an updraft
    this_tracer_ds = this_tracer_ds.where(this_tracer_ds["air_mass"] > 0)
    this_storm_stats = {}
    # print(f'{blt.STORM_LABELS[storm]}:')
    for ix, tracer in enumerate(tracer_stats_tracers):
        this_tracer_stats = {}
        this_tracer_stats["mean_mean"] = (
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .mean(...)
            .values
        )
        this_tracer_stats["max_mean"] = (
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .max(...)
            .values
        )
        this_tracer_stats["max_mean_time"] = this_tracer_ds["t_minutes"].values[
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .argmax("time")
            .values
        ]
        this_tracer_stats["max_mean_normalized"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["air_mass"]["t_minutes"]
                / this_tracer_ds["air_mass"]
            )
            .max(...)
            .values
        )
        this_tracer_stats["max_mean_normalized_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["t_minutes"]
                / this_tracer_ds["air_mass"]
            )
            .argmax("time")
            .values
        ]
        this_tracer_stats["eos_mean"] = (
            this_tracer_ds.isel({"time": -1})[tracer + "_nc"]
            / this_tracer_ds["air_mass"].isel({"time": -1})
        ).values
        this_post_100_ds = this_tracer_ds.isel({"time": slice(20, 37)})
        this_tracer_stats["max_mean_post_t100"] = (
            (this_post_100_ds[tracer + "_nc"] / this_post_100_ds["air_mass"])
            .max(...)
            .values
        )
        this_tracer_stats["min_mean_post_t100"] = (
            (this_post_100_ds[tracer + "_nc"] / this_post_100_ds["air_mass"])
            .min(...)
            .values
        )
        this_tracer_stats["mean_tracer_fraction"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["total_fs_tracer_nc"]
            )
            .mean()
            .values
        )
        this_tracer_stats["max_tracer_fraction"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["total_fs_tracer_nc"]
            )
            .max()
            .values
        )
        this_tracer_stats["max_tracer_fraction_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["total_fs_tracer_nc"]
            )
            .argmax("time")
            .values
        ]
        this_tracer_stats["max_tracer_fraction_normalized"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / (
                    this_tracer_ds["total_fs_tracer_nc"]
                    / this_tracer_ds["t_minutes"]
                )
            )
            .max()
            .values
        )
        this_tracer_stats["max_tracer_fraction_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["t_minutes"]
                / this_tracer_ds["total_fs_tracer_nc"]
            )
            .argmax("time")
            .values
        ]

        this_tracer_stats["mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .mean()
            .values
        )
        this_tracer_stats["max_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .max()
            .values
        )
        this_tracer_stats["eos_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": -1})
            .values
        )
        this_tracer_stats["integrated_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .sum()
            .values
        )
        for t_minutes in range(50, 91, 5):
            this_tracer_stats[f"t{t_minutes}_rain_production_efficiency"] = (
                (
                    this_tracer_ds["ACC" + tracer + "_nc"]
                    / this_tracer_ds["air_mass"]
                )
                .isel({"time": t_minutes // 5})
                .values
            )
        this_tracer_stats["t90_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": 90 // 5})
            .values
        )
        this_tracer_stats["t100_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": 100 // 5})
            .values
        )
        # Calculate the mean between the period of max updraft to max rain
        this_tracer_stats["t50_t70_mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": slice(50 // 5, (70 // 5) + 1)})
            .mean("time")
            .values
        )
        # Calculate the mean between the period of max updraft to max rain
        this_tracer_stats["t50_t75_mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": slice(50 // 5, (75 // 5) + 1)})
            .mean("time")
            .values
        )
        for time in [70, 75]:
            this_tracer_stats[
                f"through_t{time}_max_rain_production_efficiency"
            ] = (
                (
                    this_tracer_ds["ACC" + tracer + "_nc"]
                    / this_tracer_ds["air_mass"]
                )
                .isel({"time": slice(None, (time // 5) + 1)})
                .max("time")
                .values
            )
        this_tracer_stats["effective_rain_production_efficiency"] = (
            # this_tracer_stats["t50_t70_mean_rain_production_efficiency"]
            this_tracer_stats["t70_rain_production_efficiency"]
            if storm == "ic_wk"
            else this_tracer_stats["eos_rain_production_efficiency"]
        )
        this_tracer_stats["t180_over_t75_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            ).isel({"time": 180 // 5})
            / (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            ).isel({"time": 75 // 5})
        ).values
        this_tracer_stats["t180_over_t75_updraft_mass"] = (
            (this_tracer_ds["air_mass"]).isel({"time": 180 // 5})
            / (this_tracer_ds["air_mass"]).isel({"time": 75 // 5})
        ).values
        this_tracer_stats["max_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .max()
            .values
        )
        this_tracer_stats["max_entrainment_efficiency_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .argmax("time")
            .values
        ]
        this_tracer_stats["mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .mean()
            .values
        )
        this_tracer_stats["integrated_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .sum()
            .values
        )
        this_tracer_stats["eos_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": -1})
            .values
        )
        this_tracer_stats["t120_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(24, 37)})
            .mean()
            .values
        )
        this_tracer_stats["t150_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(150 // 5, 37)})
            .mean()
            .values
        )
        this_tracer_stats["t85_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(17, 37)})
            .mean()
            .values
        )

        this_storm_stats[tracer] = this_tracer_stats
    tracer_stats[storm] = this_storm_stats

for k, v in tracer_stats.items():
    display(HTML(f"<h2>{blt.STORM_LABELS[k]}</h2>"))
    display(pd.DataFrame(v))

blt.save_figure_data(tracer_stats, "entrainment_factor_stats")

# Also display each of them relative to the others
for comparison_pair in [
    ["sl_wk", "ic_wk"],
    ["ic_wk", "sl_wk"],
    ["sc_wk", "ic_wk"],
    ["ic_wk", "sc_wk"],
    ["sl_wk", "sc_wk"],
    ["sc_wk", "sl_wk"],
]:
    display(
        HTML(
            f"<h2>{blt.STORM_LABELS[comparison_pair[0]]} /"
            f" {blt.STORM_LABELS[comparison_pair[1]]}</h2>"
        )
    )
    display(
        pd.DataFrame(tracer_stats[comparison_pair[0]])
        / pd.DataFrame(tracer_stats[comparison_pair[1]])
    )
```

#### Identification of upper level cutoff

```python
height_cutoff_times = {"ic_wk": 100, "sl_wk": 180, "sc_wk": 180}

cm.plotting.start_figure("upper_level_height_cutoff")


fig, axs = plt.subplots(ncols=3, figsize=(10, 3))
for storm_ix, (storm, this_ul_lite_ds) in enumerate(storm_dss.items()):
    # Limit to the right time
    this_ul_lite_ds = this_ul_lite_ds.sel(
        {"time": blt.minutes_to_simulation_time(height_cutoff_times[storm])},
        drop=True,
    )
    # Look at condensate
    ax = axs[storm_ix]
    pc = ax.pcolormesh(
        this_ul_lite_ds["x"],
        this_ul_lite_ds["z"],
        this_ul_lite_ds["R_condensate"].sum("y").values,
        norm=colors.LogNorm(vmin=1e-5),
    )
    fig.colorbar(pc, ax=ax)

fig, axs = plt.subplots(ncols=3, figsize=(10, 3))
for storm_ix, (storm, this_ul_lite_ds) in enumerate(storm_dss.items()):
    this_ul_lite_ds = this_ul_lite_ds.sel(
        {"time": blt.minutes_to_simulation_time(height_cutoff_times[storm])},
        drop=True,
    )
    # Look at condensate without strong vertical motion
    this_ul_lite_ds = this_ul_lite_ds.where(
        (this_ul_lite_ds["WC"] < 0.5) & (this_ul_lite_ds["WC"] > -0.5)
    )
    ax = axs[storm_ix]
    pc = ax.pcolormesh(
        this_ul_lite_ds["x"],
        this_ul_lite_ds["z"],
        this_ul_lite_ds["R_condensate"].sum("y").values,
        norm=colors.LogNorm(vmin=1e-5, vmax=3),
    )
    fig.colorbar(pc, ax=ax)

    # Add a line for the base-state temperature profile
    t_ax = ax.twiny()
    this_t_profile = (
        storm_dss[storm].isel({"time": 0})["T"].mean(["x", "y"]).values
    )
    t_ax.plot(this_t_profile, storm_dss[storm]["z"].values, color="pink")

    # Add dashed vertical lines
    t_ax.axvline(
        273.15, color="pink", linestyle="--", linewidth=2, label="0 °C"
    )
    t_ax.axvline(
        273.15 - 38, color="pink", linestyle="--", linewidth=2, label="-38 °C"
    )
```

#### Right mover vs full domain comparison

```python
# All four columns below are now the same calculation, so they plot identically. See
# the note at the head of this section for why: the main upper-level calculation is
# no longer restricted to the supercell's right mover, so there is no longer a
# right-mover version of either quantity to contrast with the full-domain one.
#
# `this_z_ul_tracer_dfs[10]` and `ul_flux_dss[10]` are the source notebook's stale
# names for the two dicts built above.
fig, axs = plt.subplots(
    ncols=4, nrows=2, figsize=(10, 6), sharex=True, sharey="row"
)

z_km = 10


def plot_rm_comparison(axs, tracer_df, flux_ds):
    axs = axs.flatten()

    # RM normalized by RM
    axs[0].plot(
        tracer_df["t_minutes"],
        tracer_df[f"total_fs_tracer_nc"].values
        / flux_ds["cumu_upward_mass_flux"].values,
    )
    for tracer_ix, tracer in enumerate(blt.RS_TRACERS):
        # Normalized by max updraft mass
        axs[1].plot(
            tracer_df["t_minutes"],
            tracer_df[f"{tracer}_nc"].values
            / flux_ds["cumu_upward_mass_flux"].values,
            color=mpl.colormaps[blt.PROGRESSIVE_CMAP](
                tracer_ix / len(blt.RS_TRACERS)
            ),
        )


# RM tracer by RM flux
plot_rm_comparison(
    axs[:, 0],
    tracer_df=ul_tracer_dfs_all_heights[z_km]["sc_wk"],
    flux_ds=ul_flux_dss_all_heights[z_km]["sc_wk"],
)
# RM tracer by domain flux
plot_rm_comparison(
    axs[:, 1],
    tracer_df=ul_tracer_dfs_all_heights[z_km]["sc_wk"],
    flux_ds=ul_flux_ds_sc_full_domain,
)
# Domain tracer by RM flux
plot_rm_comparison(
    axs[:, 2],
    tracer_df=ul_tracer_df_sc_full_domain,
    flux_ds=ul_flux_dss_all_heights[z_km]["sc_wk"],
)
# Domain tracer by domain flux
plot_rm_comparison(
    axs[:, 3],
    tracer_df=ul_tracer_df_sc_full_domain,
    flux_ds=ul_flux_ds_sc_full_domain,
)

for ax in axs[1, :]:
    ax.set_yscale("log")
    ax.set_ylim(1, 1e5)
```

```python
# Each storm's upper-level fixed-source tracer per unit of air lofted, as a multiple of
# the isolated convection's. A ratio of ratios, so the flux estimator's bias divides out
# to the extent that it is common-mode across storms.
#
# The source notebook reads `cumu_upward_ul_mass_fluxes` here, a name that exists nowhere
# in it; the fluxes come from `ul_flux_dss_all_heights` instead.
this_z_ul_tracer_dfs = ul_tracer_dfs_all_heights[z_km]
this_z_ul_flux_dss = ul_flux_dss_all_heights[z_km]


def final_fs_tracer_per_unit_lofted_air(storm):
    return (
        this_z_ul_tracer_dfs[storm]["total_fs_tracer_nc"].iloc[-1]
        / this_z_ul_flux_dss[storm]["cumu_upward_mass_flux"].values[-1]
    )


for storm in blt.STORM_NAMES:
    print(storm)
    display(
        final_fs_tracer_per_unit_lofted_air(storm)
        / final_fs_tracer_per_unit_lofted_air("ic_wk")
    )
```

## Running it

* Use the pixi environment defined in `pixi.toml` (kernel "default"). It installs this repo's `blt_utils` package and its copy of `cloudy` in editable mode, so both import from anywhere.
* `blt_utils` holds every path. `blt.RUN_DIRS` names the run for each storm, and `blt.DERIVED_DIRS` names where that run's derived products live. To analyse different runs, edit those two dictionaries.
* Figures and the data behind them go to `analysis/paper_figures/`. The submitted manuscript's figures in `paper_figures/` are never written.
* `USE_DASK_LOCAL_CLUSTER` is `True`, because the reductions over the full output need a cluster.
* Each expensive derived product is cached on disk on first use. A cache is reused only when it covers every output time the run has on disk; otherwise it is recomputed.
* The upper-level mass flux (Figures 11 and 12) integrates the 2-second lite output. Computing it reads 5401 files per storm and takes tens of minutes per storm on the first run.

### Fixed-source tracer since rain onset

The rain-sourced tracer fraction f_RS divides in-updraft rain-sourced tracer by in-updraft fixed-source tracer. f_RS is zero by construction before rain begins. So all of its numerator's surface contact comes after rain onset. The whole-run denominator also counts surface contact from before the rain, which could never have been rainy.

The since-rain-onset version drops that earlier contact. Its rule has two parts:

1. **Rain onset** is the first output at which any 1 mm/hr rain-sourced tracer has been emitted in the storm's analysis region. Rain began after the previous output.
2. **The denominator** keeps only the fixed-source species whose emission window ends after that previous output. Each species is emitted for 5 minutes in turn, so a species records when its air was at the surface.

The same kept tracer gives a since-rain-onset q_FS. With it, q_RS = q_FS × f_RS holds exactly. The table below reports each storm's rain onset, and the time at which the first kept species began to be emitted.

```python
reload(blt)

# Fixed-source tracer emitted since each storm began to rain: the denominator of the
# since-rain-onset f_RS, and the numerator of the since-rain-onset q_FS
tracer_totals_dss = {
    storm: blt.add_fs_tracer_since_rain_onset(tracer_totals_ds)
    for storm, tracer_totals_ds in tracer_totals_dss.items()
}

# The same tracer on the in-updraft vertical profiles behind Fig. 8. The profiles carry
# the same accumulated-emission time series as the column totals, so they give the same
# rain onset and keep the same species
tracer_totals_profile_dss = {
    storm: blt.add_fs_tracer_since_rain_onset(profile_ds)
    for storm, profile_ds in tracer_totals_profile_dss.items()
}
for storm, profile_ds in tracer_totals_profile_dss.items():
    assert (
        profile_ds["fs_tracer_since_rain_onset_nc"].attrs
        == tracer_totals_dss[storm]["fs_tracer_since_rain_onset_nc"].attrs
    ), f"{storm}: the profiles and column totals give different rain onsets"

# Each storm's rain onset and the start of its kept fixed-source emission, in minutes
# since the start of the run
onset_times_df = pd.DataFrame({
    blt.STORM_LABELS[storm]: {
        "Rain onset output (min)": (
            tracer_totals_ds["fs_tracer_since_rain_onset_nc"].attrs[
                "rain_onset_minutes"
            ]
        ),
        "Kept fixed-source emission starts (min)": (
            tracer_totals_ds["fs_tracer_since_rain_onset_nc"].attrs[
                "kept_emission_start_minutes"
            ]
        ),
    }
    for storm, tracer_totals_ds in tracer_totals_dss.items()
}).T
display(onset_times_df)
```

## Figure 1: Storm soundings

```python
# Figure 1: the initial environment. One skew-T shows the thermodynamic profile,
# which all three storms share, and a column of hodographs beside it shows each
# storm's winds, which differ.
figure_name = "f01_skewt_logp"
figure_data = {}

# =============================================================================
# Plot settings
# =============================================================================
# Highest wind level on the hodographs (m), as the Fig. 1 caption states
hodograph_top_height_m = 8000
# Half-width of each hodograph's axes, and the spacing of its grid rings (m/s)
hodograph_component_range_m_per_s = 35.0
hodograph_grid_increment_m_per_s = 10
# Spacing of the skew-T's labeled pressure levels (hPa)
skewt_pressure_tick_spacing_hpa = 200
# Largest difference between two storms' initial temperature or dewpoint profiles
# (K) that still lets one skew-T stand for all three storms
shared_profile_tolerance_k = 0.05
# Gap between the hodograph column and its color bar, as a fraction of the column
# width, wide enough that the panel titles end before the color bar starts
hodograph_colorbar_pad_fraction = 0.2
# Panel titles that would run into the color bar, split over two lines
hodograph_title_overrides = {"ic_wk": "Isolated\nconvection"}

# =============================================================================
# Initial profiles
# =============================================================================
# Horizontal-mean profile of each storm at the first output time
initial_profile_dfs = {
    storm: (
        storm_ds.isel({"time": 0})
        .mean(["x", "y"])[["P", "T", "dewpoint", "UC", "VC"]]
        .to_pandas()
        .reset_index()
    )
    for storm, storm_ds in storm_dss.items()
}

# The skew-T shows only the first storm's profile, so check that the other storms
# share it
reference_storm = list(initial_profile_dfs)[0]
reference_profile_df = initial_profile_dfs[reference_storm]
for storm, profile_df in initial_profile_dfs.items():
    for variable_name in ["T", "dewpoint"]:
        largest_difference_k = np.abs(
            profile_df[variable_name].values
            - reference_profile_df[variable_name].values
        ).max()
        assert largest_difference_k < shared_profile_tolerance_k, (
            f"The {storm} initial {variable_name} profile differs from the"
            f" {reference_storm} one by {largest_difference_k:.3f} K"
        )

# Shared profile in the skew-T's units: hPa for pressure and °C for temperature
pressure_hpa = reference_profile_df["P"].values
temperature_degc = (
    (reference_profile_df["T"].values * units("K")).to("degC").magnitude
)
dewpoint_degc = (
    (reference_profile_df["dewpoint"].values * units("K")).to("degC").magnitude
)

# Path of a parcel lifted from the lowest level
surface_parcel_path_degc = (
    mpc.parcel_profile(
        pressure_hpa * units("hPa"),
        temperature_degc[0] * units("degC"),
        dewpoint_degc[0] * units("degC"),
    )
    .to("degC")
    .magnitude
)

# =============================================================================
# Layout
# =============================================================================
fig = plt.figure(figsize=(10, 7), layout="constrained")

# A wide left column for the skew-T, and a narrow right column that stacks the
# three hodographs
figure_grid = fig.add_gridspec(nrows=3, ncols=2, width_ratios=[3, 1])
# The skew-T gets its own one-cell grid within the left column. Its top margin
# then depends only on its own title, not on the two-line hodograph title
skewt_grid = figure_grid[:, 0].subgridspec(nrows=1, ncols=1)

# =============================================================================
# Skew-T
# =============================================================================
# MetPy's SkewT builds its own skewed axes, so it takes the grid slot itself
# rather than an axes made from it
skewt = SkewT(fig, rotation=30, aspect=100, subplot=skewt_grid[0, 0])
skewt_ax = skewt.ax

# Temperature, dewpoint and the surface parcel's path
temperature_line = blt.plot_and_save(
    figure_data,
    "skewt_temperature",
    skewt.plot,
    pressure_hpa,
    temperature_degc,
    "r",
    label="Temperature",
)
dewpoint_line = blt.plot_and_save(
    figure_data,
    "skewt_dewpoint",
    skewt.plot,
    pressure_hpa,
    dewpoint_degc,
    "blue",
    label="Dewpoint",
)
parcel_line = blt.plot_and_save(
    figure_data,
    "surface_parcel_path",
    skewt.plot,
    pressure_hpa,
    surface_parcel_path_degc,
    color="grey",
    linestyle="dashed",
    linewidth=2,
    label="Surface parcel path",
)

# Axis labels, pressure ticks, legend and title
skewt_ax.set_xlabel("Temperature (°C)")
skewt_ax.set_ylabel("Pressure (hPa)")
skewt_ax.yaxis.set_major_locator(
    ticker.MultipleLocator(skewt_pressure_tick_spacing_hpa)
)
skewt_ax.legend(loc="upper right")
skewt_ax.set_title(
    blt.axes_title(0, "All storms' initial thermodynamic profile")
)

# =============================================================================
# Hodographs
# =============================================================================
# One hodograph per storm, stacked in the right-hand column
hodograph_axes = [
    fig.add_subplot(figure_grid[row_index, 1]) for row_index in range(3)
]

# One height scale for all three hodographs, so a single color bar serves them all
hodograph_height_norm_km = colors.Normalize(
    vmin=0, vmax=hodograph_top_height_m / 1000
)

for storm_ix, (storm, profile_df) in enumerate(initial_profile_dfs.items()):
    hodograph_ax = hodograph_axes[storm_ix]

    # Hodograph frame with grid rings
    hodograph = Hodograph(
        hodograph_ax, component_range=hodograph_component_range_m_per_s
    )
    hodograph.add_grid(increment=hodograph_grid_increment_m_per_s)

    # Winds from the surface to the top height only, colored by height in km
    profile_below_top_df = profile_df.loc[
        profile_df["z"] <= hodograph_top_height_m
    ]
    height_colored_line = blt.plot_and_save(
        figure_data,
        f"storm-{storm}_hodograph",
        hodograph.plot_colormapped,
        profile_below_top_df["UC"].values,
        profile_below_top_df["VC"].values,
        profile_below_top_df["z"].values / 1000,
        norm=hodograph_height_norm_km,
    )

    # MetPy puts the u component on the x-axis and v on the y-axis
    hodograph_ax.set_xlabel("U (m/s)")
    hodograph_ax.set_ylabel("V (m/s)")
    hodograph_title = hodograph_title_overrides.get(
        storm, blt.STORM_LABELS[storm]
    )
    hodograph_ax.set_title(blt.axes_title(storm_ix + 1, hodograph_title))

# Height color bar on the right of the hodograph column
height_colorbar = fig.colorbar(
    height_colored_line,
    ax=hodograph_axes,
    location="right",
    pad=hodograph_colorbar_pad_fraction,
)
# Label reading top to bottom. Bottom alignment keeps the rotated label clear of the
# tick labels
height_colorbar.ax.set_ylabel("Height (km)", rotation=-90, va="bottom")

blt.save_figure_data(figure_data, figure_name)
cm.plotting.fig_multisave(fig, figure_name, blt.FIGS_DIR)
```

## Figures 11-12: Lofting to upper levels

### Setup

#### Calculate upper-level tracer concentrations

```python
# Total tracer content above the upper-level threshold height, as a time series, for
# each threshold height and storm.
#
# The supercell is limited to its right mover, as in the rest of the analysis and as
# the manuscript describes (Sect. 2.4). The other storms use their whole domain, which
# is their whole analysis region. The sensitivity tests at the end of the notebook
# compare the supercell's right mover with its full domain.
#
# Each storm's result is cached in its own upper-level directory, per height and
# region, and trusted only when it has one row per output time of the run.
ul_tracer_dfs_all_heights = {}
for z_km in blt.ALL_UL_Z_THRESHOlDS:
    this_z_ul_tracer_dfs = {}
    for storm, this_storm_ds in tqdm(storm_dss.items()):
        this_z_ul_tracer_dfs[storm] = (
            blt.read_or_compute_upper_level_tracer_dataframe(
                storm,
                this_storm_ds,
                z_km,
                right_mover_only=(storm == "sc_wk"),
            )
        )
    ul_tracer_dfs_all_heights[z_km] = this_z_ul_tracer_dfs
```

#### Calculate upper level mass fluxes

```python
# Cumulative upward and downward air mass through the upper-level threshold heights.
# The flux is integrated over every 2-second lite output time (a-L-*.h5), then sampled
# at the 5-minute analysis times.
#
# The supercell's flux is limited to its right mover, to match its upper-level tracer
# totals above. The line between the movers is applied at the flux level itself, where
# the two updrafts are still separate.
#
# Reading all 5401 lite files takes tens of minutes per storm, so each storm's result is
# cached per height and region in its own upper-level directory. Any heights missing
# from the cache are computed together, in one pass over the lite files.
ul_flux_dss_all_heights = {z_km: {} for z_km in blt.ALL_UL_Z_THRESHOlDS}
for storm, this_storm_ds in tqdm(storm_dss.items()):
    this_storm_ul_flux_dss = blt.read_or_compute_upper_level_mass_fluxes(
        storm,
        blt.ALL_UL_Z_THRESHOlDS,
        this_storm_ds["time"].values,
        right_mover_only=(storm == "sc_wk"),
    )
    for z_km, this_ul_flux_ds in this_storm_ul_flux_dss.items():
        ul_flux_dss_all_heights[z_km][storm] = this_ul_flux_ds
```

### Figure 11: Upper-level tracer amounts

```python
# Precalculate quantities we'll actually plot

# Cumulative emission of each rain-sourced tracer, the denominator of Fig. 11 (g-i). It
# is summed over each storm's analysis region, which for the supercell is its right
# mover, to match the tracer aloft and the upward mass flux. It does not depend on the
# threshold height, so it is summed once per storm.
emitted_rain_sourced_tracer_dfs = {
    storm: blt.calculate_cumulative_rain_sourced_emission(
        this_storm_ds, region_mask=this_storm_ds["in_analysis"]
    )
    for storm, this_storm_ds in tqdm(storm_dss.items())
}

for z_km, this_z_ul_tracer_dfs in ul_tracer_dfs_all_heights.items():
    for storm, this_ul_tracer_df in this_z_ul_tracer_dfs.items():
        this_cumulative_upward_mass_flux = ul_flux_dss_all_heights[z_km][storm][
            "cumu_upward_mass_flux"
        ].values

        # Fixed-source tracer above the threshold height emitted since rain onset, from
        # the same species as the in-updraft since-rain-onset q_FS
        kept_fs_species = blt.fs_species_since_rain_onset(
            tracer_totals_dss[storm]
        )["species"]
        this_ul_tracer_df["fs_tracer_since_rain_onset_nc"] = this_ul_tracer_df[
            [f"{species}_nc" for species in kept_fs_species]
        ].sum(axis="columns")

        # Times before any upward flux or emission divide zero by zero, which gives NaN.
        # The plots and the maxima skip those
        with np.errstate(invalid="ignore", divide="ignore"):
            # Fig. 11 (a-c): fixed-source tracer aloft per unit upward mass flux
            this_ul_tracer_df["ul_fs_tracer_normalized-flux"] = (
                this_ul_tracer_df["fs_tracer_since_rain_onset_nc"].values
                / this_cumulative_upward_mass_flux
            )
            for tracer in blt.RS_TRACERS:
                # Fig. 11 (d-f): rain-sourced tracer aloft per unit upward mass flux
                this_ul_tracer_df[f"ul_{tracer}_nc_normalized-flux"] = (
                    this_ul_tracer_df[f"{tracer}_nc"].values
                    / this_cumulative_upward_mass_flux
                )
                # Fig. 11 (g-i): rain-sourced tracer aloft as a fraction of the amount
                # emitted
                this_ul_tracer_df[f"ul_{tracer}_nc_normalized-acc"] = (
                    this_ul_tracer_df[f"{tracer}_nc"].values
                    / emitted_rain_sourced_tracer_dfs[storm][f"ACC{tracer}_nc"].values
                )
```

## Tracer entrainment statistics

```python
# Set the number of decimal places in the pandas display
# pd.options.display.float_format = "{:,.3f}".format
# pd.set_option('display.precision', 3)

tracer_stats = {}

# The fixed-source statistics use only fixed-source tracer emitted since rain onset,
# the mass*duration reference for the rain-sourced tracers (see the data section)
tracer_stats_tracers = blt.RS_TRACERS + ["fs_tracer_since_rain_onset"]

for storm_ix, (storm, this_tracer_ds) in enumerate(tracer_totals_dss.items()):
    # Limit to only when there's an updraft
    this_tracer_ds = this_tracer_ds.where(this_tracer_ds["air_mass"] > 0)
    this_storm_stats = {}
    # print(f'{blt.STORM_LABELS[storm]}:')
    for ix, tracer in enumerate(tracer_stats_tracers):
        this_tracer_stats = {}
        this_tracer_stats["mean_mean"] = (
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .mean(...)
            .values
        )
        this_tracer_stats["max_mean"] = (
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .max(...)
            .values
        )
        this_tracer_stats["max_mean_time"] = this_tracer_ds["t_minutes"].values[
            (this_tracer_ds[tracer + "_nc"] / this_tracer_ds["air_mass"])
            .argmax("time")
            .values
        ]
        this_tracer_stats["eos_mean"] = (
            this_tracer_ds.isel({"time": -1})[tracer + "_nc"]
            / this_tracer_ds["air_mass"].isel({"time": -1})
        ).values
        this_post_100_ds = this_tracer_ds.isel({"time": slice(20, 37)})
        this_tracer_stats["max_mean_post_t100"] = (
            (this_post_100_ds[tracer + "_nc"] / this_post_100_ds["air_mass"])
            .max(...)
            .values
        )
        this_tracer_stats["min_mean_post_t100"] = (
            (this_post_100_ds[tracer + "_nc"] / this_post_100_ds["air_mass"])
            .min(...)
            .values
        )
        # Rain-sourced tracer fraction f_RS, relative to the since-rain fixed-source
        # tracer. Its denominator is zero before rain onset, so the mean covers only
        # times after rain began.
        this_tracer_stats["mean_tracer_fraction"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["fs_tracer_since_rain_onset_nc"]
            )
            .mean()
            .values
        )
        this_tracer_stats["max_tracer_fraction"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["fs_tracer_since_rain_onset_nc"]
            )
            .max()
            .values
        )
        this_tracer_stats["max_tracer_fraction_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["fs_tracer_since_rain_onset_nc"]
            )
            .argmax("time")
            .values
        ]

        this_tracer_stats["mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .mean()
            .values
        )
        this_tracer_stats["max_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .max()
            .values
        )
        this_tracer_stats["eos_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": -1})
            .values
        )
        this_tracer_stats["integrated_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .sum()
            .values
        )
        for t_minutes in range(50, 91, 5):
            this_tracer_stats[f"t{t_minutes}_rain_production_efficiency"] = (
                (
                    this_tracer_ds["ACC" + tracer + "_nc"]
                    / this_tracer_ds["air_mass"]
                )
                .isel({"time": t_minutes // 5})
                .values
            )
        this_tracer_stats["t90_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": 90 // 5})
            .values
        )
        this_tracer_stats["t100_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": 100 // 5})
            .values
        )
        # Calculate the mean between the period of max updraft to max rain
        this_tracer_stats["t50_t70_mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": slice(50 // 5, (70 // 5) + 1)})
            .mean("time")
            .values
        )
        # Calculate the mean between the period of max updraft to max rain
        this_tracer_stats["t50_t75_mean_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            )
            .isel({"time": slice(50 // 5, (75 // 5) + 1)})
            .mean("time")
            .values
        )
        for time in [70, 75]:
            this_tracer_stats[
                f"through_t{time}_max_rain_production_efficiency"
            ] = (
                (
                    this_tracer_ds["ACC" + tracer + "_nc"]
                    / this_tracer_ds["air_mass"]
                )
                .isel({"time": slice(None, (time // 5) + 1)})
                .max("time")
                .values
            )
        this_tracer_stats["effective_rain_production_efficiency"] = (
            # this_tracer_stats["t50_t70_mean_rain_production_efficiency"]
            this_tracer_stats["t70_rain_production_efficiency"]
            if storm == "ic_wk"
            else this_tracer_stats["eos_rain_production_efficiency"]
        )
        this_tracer_stats["t180_over_t75_rain_production_efficiency"] = (
            (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            ).isel({"time": 180 // 5})
            / (
                this_tracer_ds["ACC" + tracer + "_nc"]
                / this_tracer_ds["air_mass"]
            ).isel({"time": 75 // 5})
        ).values
        this_tracer_stats["t180_over_t75_updraft_mass"] = (
            (this_tracer_ds["air_mass"]).isel({"time": 180 // 5})
            / (this_tracer_ds["air_mass"]).isel({"time": 75 // 5})
        ).values
        this_tracer_stats["max_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .max()
            .values
        )
        this_tracer_stats["max_entrainment_efficiency_time"] = this_tracer_ds[
            "t_minutes"
        ].values[
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .argmax("time")
            .values
        ]
        this_tracer_stats["mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .mean()
            .values
        )
        this_tracer_stats["integrated_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .sum()
            .values
        )
        this_tracer_stats["eos_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": -1})
            .values
        )
        this_tracer_stats["t120_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(24, 37)})
            .mean()
            .values
        )
        this_tracer_stats["t150_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(150 // 5, 37)})
            .mean()
            .values
        )
        this_tracer_stats["t85_to_t180_mean_entrainment_efficiency"] = (
            (
                this_tracer_ds[tracer + "_nc"]
                / this_tracer_ds["ACC" + tracer + "_nc"]
            )
            .isel({"time": slice(17, 37)})
            .mean()
            .values
        )

        this_storm_stats[tracer] = this_tracer_stats
    tracer_stats[storm] = this_storm_stats

for k, v in tracer_stats.items():
    display(HTML(f"<h2>{blt.STORM_LABELS[k]}</h2>"))
    display(pd.DataFrame(v))

blt.save_figure_data(tracer_stats, "entrainment_factor_stats")

# Also display each of them relative to the others
for comparison_pair in [
    ["sl_wk", "ic_wk"],
    ["ic_wk", "sl_wk"],
    ["sc_wk", "ic_wk"],
    ["ic_wk", "sc_wk"],
    ["sl_wk", "sc_wk"],
    ["sc_wk", "sl_wk"],
]:
    display(
        HTML(
            f"<h2>{blt.STORM_LABELS[comparison_pair[0]]} /"
            f" {blt.STORM_LABELS[comparison_pair[1]]}</h2>"
        )
    )
    display(
        pd.DataFrame(tracer_stats[comparison_pair[0]])
        / pd.DataFrame(tracer_stats[comparison_pair[1]])
    )
```

### Surface-air comparisons since rain onset

**Question.** Do the storms' f_RS and q_FS comparisons change when only surface contact since rain onset is counted?

**What to look for.** Compare the "Whole run" and "Since rain onset" rows of each table. In the ratio columns, a value above 1 means the first storm named has the larger peak.

```python
# Denominators of the rain-sourced tracer fraction f_RS: all in-updraft fixed-source
# tracer (whole run), or only that emitted since rain onset
frs_denominator_variables = {
    "Whole run": "total_fs_tracer_nc",
    "Since rain onset": "fs_tracer_since_rain_onset_nc",
}

# Peak f_RS over the run for each storm, denominator and rain rate. Times with no
# updraft divide zero by zero, and the maximum skips those NaNs.
peak_frs_rows = {}
with np.errstate(invalid="ignore", divide="ignore"):
    for (
        denominator_label,
        denominator_variable,
    ) in frs_denominator_variables.items():
        for tracer, threshold in zip(
            blt.RS_TRACERS, blt.RS_TRACER_THRESHOLDS_MM_HR
        ):
            peak_frs_rows[(denominator_label, f"{threshold} mm/hr")] = {
                blt.STORM_LABELS[storm]: float(
                    (
                        tracer_totals_ds[f"{tracer}_nc"]
                        / tracer_totals_ds[denominator_variable]
                    ).max()
                )
                for storm, tracer_totals_ds in tracer_totals_dss.items()
            }
peak_frs_df = pd.DataFrame(peak_frs_rows).T
peak_frs_df.index.names = ["f_RS denominator", "Rain rate threshold"]

display(HTML("<h3>Peak f_RS over the run</h3>"))
display(peak_frs_df.round(3))
```

```python
# Pairs of storms to compare, as (numerator storm, denominator storm)
storm_pairs = {
    "Squall line / Isolated convection": ("sl_wk", "ic_wk"),
    "Supercell / Squall line": ("sc_wk", "sl_wk"),
    "Supercell / Isolated convection": ("sc_wk", "ic_wk"),
}

# Each storm's peak f_RS divided by another's, for each denominator and rain rate
peak_frs_ratio_df = pd.DataFrame({
    pair_label: (
        peak_frs_df[blt.STORM_LABELS[numerator_storm]]
        / peak_frs_df[blt.STORM_LABELS[denominator_storm]]
    )
    for pair_label, (numerator_storm, denominator_storm) in storm_pairs.items()
})

display(HTML("<h3>Ratio of peak f_RS between storms</h3>"))
display(peak_frs_ratio_df.round(2))
```

```python
# Fixed-source tracer counted in each version of q_FS: all of it (whole run), or only
# that emitted since rain onset
qfs_numerator_variables = {
    "Whole run": "total_fs_tracer_nc",
    "Since rain onset": "fs_tracer_since_rain_onset_nc",
}

# Peak q_FS over the run for each storm and version, in # kg^-1. Times with no updraft
# divide zero by zero, and the maximum skips those NaNs.
with np.errstate(invalid="ignore", divide="ignore"):
    peak_qfs_df = pd.DataFrame({
        version_label: {
            blt.STORM_LABELS[storm]: float(
                (
                    tracer_totals_ds[numerator_variable]
                    / tracer_totals_ds["air_mass"]
                ).max()
            )
            for storm, tracer_totals_ds in tracer_totals_dss.items()
        }
        for version_label, numerator_variable in qfs_numerator_variables.items()
    }).T

# Each storm's peak divided by another's
peak_qfs_ratio_df = pd.DataFrame({
    pair_label: (
        peak_qfs_df[blt.STORM_LABELS[numerator_storm]]
        / peak_qfs_df[blt.STORM_LABELS[denominator_storm]]
    )
    for pair_label, (numerator_storm, denominator_storm) in storm_pairs.items()
})

display(
    HTML("<h3>Peak q_FS (# kg<sup>-1</sup>), and its ratio between storms</h3>")
)
display(
    pd.concat(
        [peak_qfs_df.map("{:.3g}".format), peak_qfs_ratio_df.round(2)], axis=1
    )
)
```

**Result.** Counting only surface contact since rain onset changes every comparison with the isolated convection. The supercell–squall line ratios barely change.

| Peak ratio | Whole run | Since rain onset |
|---|---|---|
| f_RS, squall line / isolated convection, 1–20 mm/hr | 0.92–1.22 | 0.24–0.28 |
| f_RS, supercell / squall line, 1–20 mm/hr | 2.1–2.3 | 1.7–2.4 |
| f_RS, supercell / isolated convection, 1–20 mm/hr | 2.1–2.6 | 0.47–0.58 |
| f_RS, squall line / isolated convection, 40 mm/hr | 3.8 | 0.74 |
| f_RS, supercell / squall line, 40 mm/hr | 0.29 | 0.35 |
| q_FS, squall line / isolated convection | 2.5 | 21 |
| q_FS, supercell / squall line | 0.73 | 0.82 |
| q_FS, supercell / isolated convection | 1.8 | 18 |

- **The isolated convection takes in little surface air once it rains.** Its peak q_FS falls from 8.8 × 10⁵ to 7.2 × 10⁴ # kg⁻¹ when only contact since rain onset counts. Most of the surface air in its updraft arrived before the rain.
- **The surface air it takes in after rain begins is mostly rainy.** For 1–20 mm/hr, its since-rain-onset f_RS is about 4x the squall line's and about 2x the supercell's.
- **The squall line's and supercell's f_RS rise by similar factors.** Their peaks rise 2.2x and 1.6x at 1 mm/hr, so the supercell still leads by about 2x.

**Implication.** The decomposition q_RS = q_FS × f_RS holds exactly for the since-rain-onset versions. With them, the isolated convection's low q_RS comes from its small intake of surface air after rain begins, not from dilution by non-rainy air.

#### Limiting the supercell to its right mover changes only the 40 mm/hr comparison

**Question.** The main upper-level calculation limits the supercell to its right mover (Sect. 2.4 of the manuscript). The two movers' anvils overlap aloft, so tracer lofted by one mover can drift over the other. Does that overlap bias the right mover's upper-level tracer? And how do the Sect. 3.4 comparisons differ with the supercell's full domain?

**Two ways to limit to the right mover.**

1. **Static line.** Apply the line between the movers to all three quantities. These are the upward mass flux through 10 km, the tracer above 10 km, and the surface emission. This is what Fig. 11 does. Its weakness is the tracer aloft. Right-mover anvil air that drifts north of the line is dropped, and left-mover anvil air that drifts south is kept.
2. **Attribution at the 10 km crossing.** Air is assigned to a mover where it crosses 10 km upward, inside the updrafts, which the line does separate. The quantity is the tracer carried upward through 10 km per kg of air carried upward. Spreading aloft cannot bias it. However, it counts air as it crosses upward, so it ignores air that later sinks back. It also uses the 5-minute output, the only output with tracers.

Every fixed-source quantity here counts only tracer emitted since rain onset, as elsewhere in the notebook.

**What to look for.** In the tables, compare the "Right mover" rows with the "Full domain" rows. If the two methods give similar right-mover / full-domain ratios, the static line is not biased much by the spreading.

The last table is the check that Sect. 2.4 of the manuscript describes. It compares two shares of the supercell's tracer that the right mover holds. $f_{flux}$ is its share of the tracer carried upward through 10 km. $f_N$ is its share of the tracer above 10 km at the end of the run. Equal shares mean that exchange aloft does not bias the right mover's totals.

```python
# The supercell's tracer above the threshold height over its full domain, including the
# left mover. The main calculation uses the right mover only; this is the alternative it
# is compared against here.
z_km = blt.UL_Z_THRESHOLD
ul_tracer_df_sc_full_domain = blt.read_or_compute_upper_level_tracer_dataframe(
    "sc_wk", storm_dss["sc_wk"], z_km, right_mover_only=False
)

# Fixed-source tracer aloft emitted since rain onset, from the same species as the
# in-updraft since-rain-onset q_FS
supercell_kept_fs_species = blt.fs_species_since_rain_onset(
    tracer_totals_dss["sc_wk"]
)["species"]
ul_tracer_df_sc_full_domain["fs_tracer_since_rain_onset_nc"] = (
    ul_tracer_df_sc_full_domain[
        [f"{species}_nc" for species in supercell_kept_fs_species]
    ].sum(axis="columns")
)
```

```python
# The supercell's cumulative upward mass flux through each threshold height over its
# full domain, for the comparison. Each height is cached in its own file, separate from
# the right mover's caches that the main calculation reads.
ul_flux_dss_sc_full_domain = blt.read_or_compute_upper_level_mass_fluxes(
    "sc_wk",
    blt.ALL_UL_Z_THRESHOlDS,
    storm_dss["sc_wk"]["time"].values,
    right_mover_only=False,
)
```

```python
# Cumulative surface emission of each rain-sourced tracer, the denominator of Fig. 11
# (g-i), for each version compared here. The squall line's and the supercell right
# mover's come from the main calculation. The supercell's full domain is summed here.
emitted_by_version = {
    "Squall line": emitted_rain_sourced_tracer_dfs["sl_wk"],
    "Supercell, full domain": blt.calculate_cumulative_rain_sourced_emission(
        storm_dss["sc_wk"]
    ),
    "Supercell, right mover": emitted_rain_sourced_tracer_dfs["sc_wk"],
}
```

```python
def normalized_upper_level_tracer(tracer_df, upward_mass_flux, emitted_df):
    """Fig. 11's three quantities for one storm or one region of a storm.

    Mechanism: the tracer totals above the threshold height are divided by the
    cumulative upward mass flux through it (Fig. 11 rows 1 and 2). The
    fixed-source total counts only tracer emitted since rain onset, so
    ``tracer_df`` must hold ``fs_tracer_since_rain_onset_nc``. Each
    rain-sourced tracer's total is also divided by its cumulative surface emission,
    from ``blt.calculate_cumulative_rain_sourced_emission`` (row 3). Times before
    any upward flux or emission divide zero by zero, which gives NaN; the plots and
    the maxima skip those.

    Returns:
        A DataFrame indexed by minutes since the start, with columns
        ``fs_per_flux``, ``<tracer>_per_flux`` and ``<tracer>_per_emitted``.
    """
    with np.errstate(invalid="ignore", divide="ignore"):
        normalized_columns = {
            "fs_per_flux": (
                tracer_df["fs_tracer_since_rain_onset_nc"].values
                / upward_mass_flux
            )
        }
        for tracer in blt.RS_TRACERS:
            normalized_columns[f"{tracer}_per_flux"] = (
                tracer_df[f"{tracer}_nc"].values / upward_mass_flux
            )
            normalized_columns[f"{tracer}_per_emitted"] = (
                tracer_df[f"{tracer}_nc"].values
                / emitted_df[f"ACC{tracer}_nc"].values
            )
    return pd.DataFrame(normalized_columns, index=tracer_df["t_minutes"].values)


# Static-line version of Fig. 11's quantities: the squall line for reference, and the
# supercell over its full domain and over its right mover (as in Fig. 11)
z_km = blt.UL_Z_THRESHOLD
normalized_by_version = {
    "Squall line": normalized_upper_level_tracer(
        ul_tracer_dfs_all_heights[z_km]["sl_wk"],
        ul_flux_dss_all_heights[z_km]["sl_wk"]["cumu_upward_mass_flux"].values,
        emitted_by_version["Squall line"],
    ),
    "Supercell, full domain": normalized_upper_level_tracer(
        ul_tracer_df_sc_full_domain,
        ul_flux_dss_sc_full_domain[z_km]["cumu_upward_mass_flux"].values,
        emitted_by_version["Supercell, full domain"],
    ),
    "Supercell, right mover": normalized_upper_level_tracer(
        ul_tracer_dfs_all_heights[z_km]["sc_wk"],
        ul_flux_dss_all_heights[z_km]["sc_wk"]["cumu_upward_mass_flux"].values,
        emitted_by_version["Supercell, right mover"],
    ),
}
```

```python
# Fig. 11's supercell column twice, over the full domain and over the right mover only
# (static line), with the squall line's column for reference
fig, axs = plt.subplots(
    nrows=3, ncols=3, figsize=(10, 10), sharex=True, sharey="row"
)
# Room for the figure title at the top and the tracer legend at the bottom
fig.subplots_adjust(top=0.9, bottom=0.12, hspace=0.35)

for column_ix, (version_label, normalized_df) in enumerate(
    normalized_by_version.items()
):
    # Row 1: fixed-source tracer aloft per unit upward mass flux
    axs[0, column_ix].plot(normalized_df.index, normalized_df["fs_per_flux"])
    for tracer_ix, tracer in enumerate(blt.RS_TRACERS):
        tracer_color = mpl.colormaps[blt.PROGRESSIVE_CMAP](
            tracer_ix / len(blt.RS_TRACERS)
        )
        tracer_label = f"{blt.RS_TRACER_THRESHOLDS_MM_HR[tracer_ix]} mm/hr"
        # Row 2: rain-sourced tracer aloft per unit upward mass flux
        axs[1, column_ix].plot(
            normalized_df.index,
            normalized_df[f"{tracer}_per_flux"],
            color=tracer_color,
            label=tracer_label,
        )
        # Row 3: rain-sourced tracer aloft as a fraction of the amount emitted
        axs[2, column_ix].plot(
            normalized_df.index,
            normalized_df[f"{tracer}_per_emitted"],
            color=tracer_color,
        )
    for row_ix in range(3):
        axs[row_ix, column_ix].set_title(
            blt.axes_title(3 * row_ix + column_ix, version_label)
        )

# Axis labels and scales, matching Fig. 11
axs[0, 0].set_ylabel(r"$N_{FS-UL}/F_{upward}$ (# kg$^{-1}$)")
axs[1, 0].set_ylabel(r"$N_{RS-UL}/F_{upward}$ (# kg$^{-1}$)")
axs[2, 0].set_ylabel(r"$N_{RS-UL}/E_{RS}$ (fraction)")
axs[1, 0].set_yscale("log")
axs[1, 0].set_ylim([1e-1, 1e5])
for ax in axs[-1, :]:
    blt.label_x_axis(ax)
# One legend for the rain-sourced tracers, below the bottom row
fig.legend(
    *axs[1, 0].get_legend_handles_labels(),
    loc="lower center",
    bbox_to_anchor=(0.5, -0.01),
    ncol=5,
)
fig.suptitle(
    "Upper-level tracer at z ≥ 10 km: supercell full domain vs right mover"
)
```

```python
# The supercell / squall line ratios that Sect. 3.4 of the manuscript quotes, for each
# supercell version: the supercell's value at the end of the run divided by the squall
# line's, per tracer and normalization
squall_line_df = normalized_by_version["Squall line"]
ratio_rows = {}
for version_label in ["Supercell, full domain", "Supercell, right mover"]:
    supercell_df = normalized_by_version[version_label]
    for normalization in ["per_flux", "per_emitted"]:
        ratio_rows[(version_label, normalization)] = {
            f"{threshold} mm/hr": (
                supercell_df[f"{tracer}_{normalization}"].iloc[-1]
                / squall_line_df[f"{tracer}_{normalization}"].iloc[-1]
            )
            for tracer, threshold in zip(
                blt.RS_TRACERS, blt.RS_TRACER_THRESHOLDS_MM_HR
            )
        }
supercell_over_squall_line_df = pd.DataFrame(ratio_rows).T
supercell_over_squall_line_df.index.names = [
    "Supercell version",
    "Normalization",
]
display(
    HTML(
        "<h3>Supercell / squall line rain-sourced tracer aloft at the end of the"
        " run, static line</h3>"
    )
)
display(supercell_over_squall_line_df.round(2))

# Fixed-source tracer aloft per unit upward flux, as Sect. 3.4 compares it: the
# supercell's end-of-run value against the squall line's end-of-run value and the
# isolated storm's peak
isolated_storm_fs_per_flux = (
    ul_tracer_dfs_all_heights[z_km]["ic_wk"][
        "fs_tracer_since_rain_onset_nc"
    ].values
    / ul_flux_dss_all_heights[z_km]["ic_wk"]["cumu_upward_mass_flux"].values
)
fixed_source_ratio_df = pd.DataFrame({
    version_label: {
        "SC end / SL end": (
            normalized_by_version[version_label]["fs_per_flux"].iloc[-1]
            / squall_line_df["fs_per_flux"].iloc[-1]
        ),
        "SC end / IC peak": (
            normalized_by_version[version_label]["fs_per_flux"].iloc[-1]
            / np.nanmax(isolated_storm_fs_per_flux)
        ),
    }
    for version_label in ["Supercell, full domain", "Supercell, right mover"]
}).T
display(
    HTML("<h3>Fixed-source tracer aloft per unit upward flux, static line</h3>")
)
display(fixed_source_ratio_df.round(2))
```

```python
# Attribution at the 10 km crossing: tracer carried upward through the threshold level
# per kg of air carried upward, accumulated over the run. The supercell's region is
# decided where the air crosses the level, inside the updrafts, so the spreading of
# the two anvils above it cannot move tracer between the movers.
upward_transport_dfs = {
    "Squall line": blt.read_or_cache_pickle(
        blt.UPPER_LEVEL_DERIVED_DIRS["sl_wk"]
        / f"upward_transport_by_species_z-{z_km * 1000}.pkl",
        blt.calculate_upward_transport_through_level,
        storm_dss["sl_wk"],
        z_km * 1000,
        expected_time_count=storm_dss["sl_wk"].sizes["time"],
    ),
    "Supercell, full domain": blt.read_or_cache_pickle(
        blt.UPPER_LEVEL_DERIVED_DIRS["sc_wk"]
        / f"upward_transport_by_species_z-{z_km * 1000}.pkl",
        blt.calculate_upward_transport_through_level,
        storm_dss["sc_wk"],
        z_km * 1000,
        expected_time_count=storm_dss["sc_wk"].sizes["time"],
    ),
    "Supercell, right mover": blt.read_or_cache_pickle(
        blt.UPPER_LEVEL_DERIVED_DIRS["sc_wk"]
        / f"upward_transport_by_species_right_mover_z-{z_km * 1000}.pkl",
        blt.calculate_upward_transport_through_level,
        storm_dss["sc_wk"],
        z_km * 1000,
        horizontal_mask=storm_dss["sc_wk"]["in_analysis"],
        expected_time_count=storm_dss["sc_wk"].sizes["time"],
    ),
}

# Fixed-source tracer carried upward that was emitted since rain onset, from the same
# species as the in-updraft since-rain-onset q_FS. Both supercell versions use the
# supercell's species.
transport_storms = {
    "Squall line": "sl_wk",
    "Supercell, full domain": "sc_wk",
    "Supercell, right mover": "sc_wk",
}
for version_label, transport_df in upward_transport_dfs.items():
    kept_fs_species = blt.fs_species_since_rain_onset(
        tracer_totals_dss[transport_storms[version_label]]
    )["species"]
    transport_df["cumu_upward_fs_tracer_since_rain_onset_nc"] = transport_df[
        [f"cumu_upward_{species}_nc" for species in kept_fs_species]
    ].sum(axis="columns")

# Tracer carried upward per kg of air carried upward, at the end of the run (# kg^-1)
transported_tracer_labels = {"fs_tracer_since_rain_onset": "fixed-source"} | {
    tracer: f"{threshold} mm/hr"
    for tracer, threshold in zip(blt.RS_TRACERS, blt.RS_TRACER_THRESHOLDS_MM_HR)
}
transport_per_kg_df = pd.DataFrame({
    version_label: {
        tracer_label: (
            transport_df[f"cumu_upward_{tracer}_nc"].iloc[-1]
            / transport_df["cumu_upward_air_mass"].iloc[-1]
        )
        for tracer, tracer_label in transported_tracer_labels.items()
    }
    for version_label, transport_df in upward_transport_dfs.items()
}).T
display(
    HTML(
        "<h3>Tracer carried upward through 10 km per kg of air (#"
        " kg<sup>-1</sup>)</h3>"
    )
)
display(transport_per_kg_df)
```

```python
# Do the two methods agree? Right mover / full domain at the end of the run, for the
# static line (tracer aloft per unit upward mass flux) and for attribution at the
# crossing (tracer carried upward per unit air carried upward). Both are per unit
# upward air mass, so each is the right mover's share of the tracer divided by its
# share of the upward air. The next cell compares the tracer shares themselves
static_line_df = normalized_by_version["Supercell, right mover"].iloc[-1] / (
    normalized_by_version["Supercell, full domain"].iloc[-1]
)
method_comparison_df = pd.DataFrame({
    "Static line": {
        tracer_label: static_line_df[
            "fs_per_flux"
            if tracer == "fs_tracer_since_rain_onset"
            else f"{tracer}_per_flux"
        ]
        for tracer, tracer_label in transported_tracer_labels.items()
    },
    "Attribution at 10 km": (
        transport_per_kg_df.loc["Supercell, right mover"]
        / transport_per_kg_df.loc["Supercell, full domain"]
    ),
}).T
display(
    HTML(
        "<h3>Right mover / full domain, tracer per unit upward air mass, end of"
        " run</h3>"
    )
)
display(method_comparison_df.round(2))

# Supercell / squall line from attribution at the crossing, for comparison with the
# static-line table above
display(
    HTML("<h3>Supercell / squall line, attribution at 10 km, end of run</h3>")
)
display(
    (
        transport_per_kg_df.loc[
            ["Supercell, full domain", "Supercell, right mover"]
        ]
        / transport_per_kg_df.loc["Squall line"]
    ).round(2)
)
```

```python
# The check that Sect. 2.4 of the manuscript describes. f_flux is the right mover's share
# of the supercell's tracer carried upward through 10 km over the run. f_N is its share
# of the supercell's tracer above 10 km at the end of the run. Equal shares mean that
# exchange between the movers' anvils aloft does not bias the right mover's totals
right_mover_share_df = pd.DataFrame({
    tracer_label: {
        "f_flux": (
            upward_transport_dfs["Supercell, right mover"][
                f"cumu_upward_{tracer}_nc"
            ].iloc[-1]
            / upward_transport_dfs["Supercell, full domain"][
                f"cumu_upward_{tracer}_nc"
            ].iloc[-1]
        ),
        "f_N": (
            ul_tracer_dfs_all_heights[z_km]["sc_wk"][f"{tracer}_nc"].iloc[-1]
            / ul_tracer_df_sc_full_domain[f"{tracer}_nc"].iloc[-1]
        ),
    }
    for tracer, tracer_label in transported_tracer_labels.items()
}).T

# Relative difference between the two shares, as a fraction of f_flux
right_mover_share_df["f_N / f_flux - 1"] = (
    right_mover_share_df["f_N"] / right_mover_share_df["f_flux"] - 1
)
display(
    HTML("<h3>Right mover's share of the supercell's tracer, end of run</h3>")
)
display(right_mover_share_df.round(3))
```

|---|---|---|---|---|---|---|

- **Fixed-source tracer, emitted since rain onset.** The supercell's end-of-run value is 6.8 times the squall line's with either domain. It is about 1,650 times the isolated storm's peak with either domain. The isolated storm lofts almost no surface air from after its rain began.
- **1–20 mm/hr.** The right mover alone leads the squall line by more than the full domain does.
- **40 mm/hr.** Per unit upward flux, the supercell's lead falls from 4.2 with the full domain to 1.4 with the right mover. Per unit emitted, it falls from 3.6 to 2.4. The right mover's ratio per unit emitted peaks at 4.2 before the end of the run.
- **The two methods agree.** Both give the right mover / full domain ratio of tracer per unit upward air mass. The two ratios differ by at most 0.05, for every tracer. So the spreading of the anvils aloft does not noticeably bias the static line.
- **The Sect. 2.4 check agrees within 6%.** $f_{flux}$ and $f_N$ agree within 3% for the fixed-source tracer and the 1–20 mm/hr tracers. At 40 mm/hr they are 0.223 and 0.237, which differ by 6%.
- **The left mover's updraft lofts most of the heavy-rain air.** At the end of the run, about a quarter of the supercell's 40 mm/hr tracer above 10 km lies over the right mover.

**Implication.** Fig. 11 and Sect. 3.4 use the right mover. The supercell then leads the squall line at every threshold with both normalizations. At 40 mm/hr, its lead per unit upward flux is only 1.4, so claims about the transport of rainy air need that exception.
